```python
import pandas as pd

# Verifica se 'base' DataFrame existe e não é None, se não, recarrega-o
if 'base' not in locals() or not isinstance(base, pd.DataFrame):
    print("DataFrame 'base' não encontrado ou não é um DataFrame. Recarregando dados...")
    base = pd.read_csv("https://raw.githubusercontent.com/guilhermeonrails/data-jobs/refs/heads/main/salaries.csv")

colunas_para_traduzir = {
    'ano': 'ano',
    'nivel_experiencia': 'senioridade',
    'tipo_emprego': 'contrato',
    'cargo': 'cargo',
    'salario': 'salario',
    'moeda_salario': 'moeda',
    'salario_em_usd': 'usd',
    'redidencia_empregado': 'residencia',
    'taxa_remoto': 'remoto',
    'localizacao_empresa': 'empresa',
    'tamanho_empresa': 'tamanho_empresa'
}

# Renomeia as colunas e atribui o resultado de volta a 'base'
base.rename(columns=colunas_para_traduzir, inplace=True)
print('As colunas do DataFrame foram traduzidas para:')
print(base.columns)
```

```python
import numpy as np
import pandas as pd

base_salarios = pd.DataFrame({
    'nome': ['João', 'Carlos', 'Joana', 'Paula', 'Junio'],
    'salario': [2400, 5000, np.nan, 3000, np.nan]
})
#calcula a média salarial e substitui os nulos pela média e arredonnda os valores
base_salarios['salario_media'] = base_salarios['salario'].fillna(base_salarios['salario'].mean().round(2))
base_salarios['salario_mediana'] = base_salarios['salario'].fillna(base_salarios['salario'].median())
print(base_salarios)
```

```python
base_temperatura = pd.DataFrame({
    'Dia': ['Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta', 'Sabdo', 'Domingo'],
    'Temperatura': [28, 30, 32, np.nan, 25, 17, np.nan]
})

# A coluna 'Preenchido_ffill' ainda não existe, então o ffill() deve ser aplicado em uma coluna existente.
# Assumindo que a intenção é preencher os NaNs da coluna 'Temperatura' e criar uma nova coluna com o resultado.
base_temperatura['Preenchido_ffill'] = base_temperatura['Temperatura'].ffill()
base_temperatura['Preenchido_bfill'] = base_temperatura['Temperatura'].bfill()
print(base_temperatura)
```

```python
base_cidades = pd.DataFrame({
    'nome': ['João', 'Carlos', 'Joana', 'Paula', 'Junio'],
    'cidade': ['São Paulo', np.nan, 'Belo Horizonte', np.nan, 'Rio de Janeiro']
})

base_cidades['cidade_preenchida'] = base_cidades['cidade'].fillna('Não informado')
print(base_cidades)
```

```python
import plotly.express as px

# Calcula a média salarial por senioridade
media_salarial_senioridade = base_limpo.groupby('senioridade')['usd'].mean().sort_values(ascending=False).reset_index()

# Define a ordem das categorias para o gráfico
ordem_senioridade_plot = media_salarial_senioridade['senioridade'].tolist()

# Cria o gráfico de barras com Plotly
fig = px.bar(
    media_salarial_senioridade,
    x='senioridade',
    y='usd',
    title='Média Salarial por Senioridade',
    labels={'senioridade': 'Nível de Senioridade', 'usd': 'Média Salarial em USD'},
    category_orders={'senioridade': ordem_senioridade_plot} # Aplica a ordem calculada
)

fig.show()
```

```python
#Grafico de pizza
remoto_contagem = base_limpo['remoto'].value_counts().reset_index()
remoto_contagem.columns = ['trabalho_remoto', 'quantidade']

fig = px.pie(remoto_contagem,
    names='trabalho_remoto',
    values='quantidade',

    title='Proporção dos tipos de trabalho',
    hole=0.2,
)
fig.show()
```
